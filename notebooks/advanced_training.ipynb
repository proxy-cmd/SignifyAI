# SignifyAI Advanced Training (AutoML)

This notebook trains the best available model from multiple candidates and saves model + metrics artifacts.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve().parents[0]
SRC = ROOT / 'src'
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

ROOT
```

```python
from signifyai.dataset import load_dataset
from signifyai.automl import train_automl, save_automl_outputs
from signifyai.config import (
    DEFAULT_DATASET_PATH,
    DEFAULT_MODEL_PATH,
    DEFAULT_LABELS_PATH,
    DEFAULT_METADATA_PATH,
    DEFAULT_CONFUSION_CSV_PATH,
)

dataset_path = DEFAULT_DATASET_PATH
ds = load_dataset(dataset_path)
print('Samples:', ds.x.shape[0], 'Features:', ds.x.shape[1], 'Classes:', len(set(ds.y.tolist())))
```

```python
model, result, cm = train_automl(ds.x, ds.y, use_augmentation=True)
print('Best model:', result.best_name)
print('CV macro F1:', round(result.cv_f1_macro, 4))
print('Test accuracy:', round(result.test_accuracy, 4))
print('Test macro F1:', round(result.test_f1_macro, 4))
print(result.report)
```

```python
save_automl_outputs(
    model=model,
    result=result,
    confusion=cm,
    model_path=DEFAULT_MODEL_PATH,
    labels_path=DEFAULT_LABELS_PATH,
    metadata_path=DEFAULT_METADATA_PATH,
    confusion_csv=DEFAULT_CONFUSION_CSV_PATH,
)
print('Saved:', DEFAULT_MODEL_PATH)
print('Saved:', DEFAULT_LABELS_PATH)
print('Saved:', DEFAULT_METADATA_PATH)
print('Saved:', DEFAULT_CONFUSION_CSV_PATH)
```
