## 2) Import Kaggle dataset (optional)
Set your dataset slug below.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve() / 'src'))

from signifyai.external_data import import_from_kaggle

slug = 'grassknoted/asl-alphabet'
target = Path('..') / 'data' / 'raw' / 'images'
import_from_kaggle(slug, target, force=False)
print('Imported to', target)
```

## 3) Build landmark CSV from images

```python
from signifyai.image_dataset import BuildImageDatasetConfig, build_dataset_from_images

cfg = BuildImageDatasetConfig(
    root_dir=Path('..') / 'data' / 'raw' / 'images',
    out_csv=Path('..') / 'data' / 'processed' / 'dataset.csv',
    max_images_per_class=1500,
    min_detection_confidence=0.55,
)
total, saved = build_dataset_from_images(cfg)
print('Processed images:', total)
print('Saved samples:', saved)
```

## 4) Train model

```python
from signifyai.train import TrainConfig, run_training

train_cfg = TrainConfig(
    dataset_csv=Path('..') / 'data' / 'processed' / 'dataset.csv',
    model_path=Path('..') / 'models' / 'gesture_model.joblib',
    labels_path=Path('..') / 'models' / 'labels.json',
    metadata_path=Path('..') / 'models' / 'model_metadata.json',
    calibrate_probs=True,
)
acc = run_training(train_cfg)
print('Final accuracy:', acc)
```

## 5) Run app
After training, run in terminal:

`python -u .\src\main.py`
